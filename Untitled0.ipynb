```python
a = 10
print("Integer:", a)
print("Type:", type(a))
print("Square:", a ** 2)

# Float
b = 10.5
print("\nFloat:", b)
print("Type:", type(b))
print("Rounded value:", round(b))

# Complex
c = 3 + 4j
print("\nComplex:", c)
print("Type:", type(c))
print("Real part:", c.real)
print("Imaginary part:", c.imag)

# String
name = "Python"
print("\nString:", name)
print("Type:", type(name))
print("Length:", len(name))
print("Uppercase:", name.upper())

# List
my_list = [10, 20, 30, 40]
print("\nList:", my_list)
print("Type:", type(my_list))
print("Length:", len(my_list))
my_list.append(50)
print("After append:", my_list)

# Tuple
my_tuple = (10, 20, 30)
print("\nTuple:", my_tuple)
print("Type:", type(my_tuple))
print("Length:", len(my_tuple))

# Dictionary
my_dict = {"Name": "Pranita", "Age": 18}
print("\nDictionary:", my_dict)
print("Type:", type(my_dict))
print("Keys:", my_dict.keys())
print("Values:", my_dict.values())

# Set (duplicate values are automatically removed)
my_set = {10, 20, 30, 30}
print("\nSet:", my_set)
print("Type:", type(my_set))
print("Length:", len(my_set))

# Frozenset
my_frozenset = frozenset([10, 20, 30])
print("\nFrozenset:", my_frozenset)
print("Type:", type(my_frozenset))

# Boolean
x = True
print("\nBoolean:", x)
print("Type:", type(x))

# NoneType
y = None
print("\nNoneType:", y)
print("Type:", type(y))

# Built-in functions
numbers = [10, 20, 30, 40, 50]
print("\n--- Built-in Functions ---")
print("Maximum:", max(numbers))
print("Minimum:", min(numbers))
print("Sum:", sum(numbers))
print("Length:", len(numbers))

# Math functions
import math
print("\n--- Math Functions ---")
print("Square root of 25:", math.sqrt(25))
print("Power of 2^3:", math.pow(2, 3))
print("Factorial of 5:", math.factorial(5))
print("Ceiling of 4.3:", math.ceil(4.3))
print("Floor of 4.8:", math.floor(4.8))
```

```python
df = pd.DataFrame({
    "sr.no": [1, 2, 3, 4, 5],
    "name": ["Anjali", "sara", "Amit", "shrushti", "siddhi"],
    "score": [88.0, 89.0, 90.0, np.nan, np.nan],
    "department": ["CE", "IT", "AIDS", "ECE", "IT"]
})

print("\nDataFrame:\n", df)
print("Missing values:\n", df.isnull())
print("Fill missing values with 88:\n", df.fillna(88))
print("Drop rows with missing values:\n", df.dropna())
print("Rows with score > 80:\n", df[df["score"] > 80])
print("Name and score columns:\n", df[["name", "score"]])
print("Sort by score descending:\n", df.sort_values("score", ascending=False))
print("Mean score by department:\n", df.groupby("department")["score"].mean())
print("Drop rows with missing names:\n", df.dropna(subset=["name"]))
print("Drop duplicate rows:\n", df.drop_duplicates())
print("Rename score column to marks:\n", df.rename(columns={"score": "marks"}))
print("Mean and maximum score by department:\n",
      df.groupby("department").agg({"score": ["mean", "max"]}))
print("Sort by department and score:\n",
      df.sort_values(["department", "score"], ascending=[True, False]))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Import Required Libraries
# The libraries are imported above.
```

```python
# 2. Load the Titanic Dataset
df = sns.load_dataset("titanic")
```

```python
# 3. Understand the Dataset
print("First five records:")
print(df.head())

print("\nShape of dataset:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nDataset information:")
df.info()
```

```python

# 4. Summary Statistics
print("\nSummary statistics:")
print(df.describe())
```

```python
# 5. Detect Missing Values
missing_values = df.isnull().sum()
print("\nMissing values in each column:")
print(missing_values)
```

```python
# 6. Visualize Missing Values
# Show only columns that contain missing values.
missing_values = df.isnull().sum()
missing_values = missing_values[missing_values > 0]
plt.figure(figsize=(10, 5))
sns.barplot(x=missing_values.index, y=missing_values.values)
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.title("Missing Values in Titanic Dataset")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python

# 7. Correlation Matrix
correlation = df.corr(numeric_only=True)
print("\nCorrelation matrix:")
print(correlation)
```

```python
# 8. Correlation Heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5,
    fmt=".2f"
)
plt.title("Correlation Matrix - Titanic Dataset")
plt.tight_layout()
plt.show()
```
